## What the model outputs

Logistic regression outputs a **probability** $P(\text{Churn}=1 \mid x)$ via the sigmoid of a linear score.  
A hard label requires a **threshold** (default 0.5) — that choice is a business decision (Phase 5–6).

## Why binary cross-entropy / log loss?

- Matches the probabilistic model: penalizes confident wrong probabilities heavily.
- Differentiable and convex (with the usual LR setup) → reliable optimization.
- Encourages well-ranked and better-calibrated scores than accuracy-only training.

## Training loss vs evaluation metrics

| Concept | Role |
|---|---|
| **Log loss (training objective)** | What the optimizer minimizes on the training set |
| **Holdout log loss** | Generalization of probability quality |
| **Precision / Recall / F1** | Thresholded decision quality |
| **ROC-AUC / PR-AUC** | Ranking quality across thresholds |

A model can have strong ROC-AUC but poor F1 at 0.5 if the operating point is wrong — or low log loss with mediocre accuracy if probabilities are good but the threshold is mismatched.
