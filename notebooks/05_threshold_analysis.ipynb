## Why 0.5 is not inherently optimal

- 0.5 is a convenient default when class costs are equal and the model is well calibrated.
- With ~27% churn and asymmetric business costs, you often want a **lower** threshold (more recall) or a **higher** one (fewer false alarms).
- The "best" threshold depends on whether you optimize F1, precision at a recall floor, or **expected business value** (Phase 6).
- Always report the operating point with the metrics — metrics without a threshold are incomplete for deployment.
