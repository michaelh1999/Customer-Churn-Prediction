## What "better than trivial" means

- **Majority-class**: always predict No churn. High accuracy (~73%) but **zero recall** for churners — useless for retention outreach.
- **Simple rule**: flag month-to-month customers with above-median monthly charges. Lower accuracy is acceptable if recall/F1 improve.

Any serious model must beat the majority baseline on **recall and F1** (and later log loss / PR-AUC), not just accuracy.
