## Selecting a threshold

- **0.5**: default ML operating point — not tied to costs.
- **F1-optimal**: balances precision and recall equally — still an ML heuristic.
- **Business-value-optimal**: maximizes `TP * retention_value - FP * fp_cost - FN * fn_cost` under the assumptions in `configs/default.yaml`.

When false negatives (lost customers) are much more expensive than false positives (wasted offers), the business-optimal threshold is typically **lower** than 0.5, accepting more outreach to catch more churners.
