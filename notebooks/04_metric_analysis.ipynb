## Situations to notice in this project

### 1. Accuracy improves but recall decreases
The **majority-class** baseline has high accuracy and **zero recall**. Logistic regression may show similar or even slightly lower accuracy while capturing most churners (high recall). Accuracy alone would wrongly favor the baseline.

### 2. Precision and recall trade off
With `class_weight='balanced'`, logistic regression pushes toward higher recall (catch churners) at the cost of more false positives (lower precision). Raising the decision threshold later will reverse that tradeoff.

### 3. ROC-AUC vs PR-AUC
ROC-AUC can look strong (~0.84) even when positives are only ~27% of customers. **PR-AUC** focuses on the positive class and is usually more informative for rare/costly events like churn outreach lists.

### 4. Similar accuracy, different log loss
Two models can hard-classify similarly near 0.5 yet differ in how sharp/calibrated their probabilities are. **Log loss** (and Brier) expose that gap — critical if you use probabilities for ranking or expected value.

**Takeaway:** pick metrics that match the decision (find churners, control outreach cost, rank risk), not a single leaderboard number.
